```python
a=int(input("Enter the number of student"))
student=[]
for i in range(a):
  name=input("Enter the name of student ")
  marks=int(input("Enter the Totalmarks out of 100 "))
  if marks>100 | marks<0:
    print("maths nhi aate kya..sahi se marks dal")
    continue
  else:
    student.append([name,marks])

for i in range(a):
  print(f"student {i} info:\n name:{student[i][0]} \n marks:{student[i][1]}")


''' to find marks of specific student '''
searchbyname=input("Enter the name of student to find thier marks ")

for i in range(a):
  if searchbyname==student[i][0]:
    print(f"{searchbyname} has scored {student[i][1]}")



```

```python
average=0
highestscore=0
lowestscore=100
total=0
totalpassed=0
totalfailed=0
for i  in range(a):
 total +=student[i][1]
 if student[i][1]>highestscore:
  highestscore=student[i][1]
 if student[i][1]<lowestscore:
  lowestscore=student[i][1]
  if student[i][1]>=40:
    totalpassed +=1
  else:
      totalfailed +=1

average=total/a

print(f"The highest marks scored is: {highestscore}")
print(f"The lowest marks scored is: {lowestscore} ")
print(f"Average makrs is: {average}")
print(f"total passed is {totalpassed}")
print(f"total failed is {totalfailed}")








```

```python
def grade(x,b):
  name=x
  marks=b
  if marks<=100 and marks>=90:
    print(f"{name} scored {marks}  marks")
  elif marks<90 and marks>=75:
      print(f"{name} scored {marks}  marks")
  elif marks<75 and marks>=60:
      print(f"{name} scored {marks}  marks")
  elif marks<60 and marks>=40:
      print(f"{name} scored {marks}  marks")
  else:
      print(f"{name} Failed ...better luck next time...soory to hear that")

for i in range(a):
  grade(student[i][0],student[i][1])
```
